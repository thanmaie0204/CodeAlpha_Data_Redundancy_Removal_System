```python
import pandas as pd
import sqlite3
import os
```

```python
existing_data = pd.DataFrame({
    "name": ["Rahul", "Priya", "Arjun", "Sneha", "Kiran"],
    "email": [
        "rahul@gmail.com",
        "priya@gmail.com",
        "arjun@gmail.com",
        "sneha@gmail.com",
        "kiran@gmail.com"
    ],
    "city": ["Hyderabad", "Warangal", "Khammam", "Hyderabad", "Warangal"],
    "age": [22, 21, 23, 22, 24]
})

existing_data
```

```python
new_data = pd.DataFrame({
    "name": ["Rahul", "Anu", "Priya", "Vamsi", "Kiran", "Meena"],
    "email": [
        "rahul@gmail.com",
        "anu@gmail.com",
        "priya@gmail.com",
        "vamsi@gmail.com",
        "kiran@gmail.com",
        "meena@gmail.com"
    ],
    "city": [
        "Hyderabad",
        "Khammam",
        "Warangal",
        "Hyderabad",
        "Warangal",
        "Khammam"
    ],
    "age": [22, 20, 21, 23, 24, 22]
})

new_data
```

```python
print("Existing data:")
print(existing_data)

print("\nNew incoming data:")
print(new_data)
```

```python
print("Missing values in existing data:")
print(existing_data.isnull().sum())

print("\nMissing values in new data:")
print(new_data.isnull().sum())
```

```python
new_data = new_data.dropna()

print("New data after validation:")
print(new_data)
```

```python
before = len(new_data)

new_data = new_data.drop_duplicates()

after = len(new_data)

print("Records before duplicate removal:", before)
print("Records after duplicate removal:", after)
print("Duplicates removed:", before - after)
```

```python
duplicate_mask = new_data["email"].isin(existing_data["email"])

duplicate_data = new_data[duplicate_mask]

duplicate_data
```

```python
print("Duplicate / Redundant Records:")
print(duplicate_data)
```

```python
unique_data = new_data[~duplicate_mask]

print("Unique Records:")
print(unique_data)
```

```python
print("Total incoming records:", len(new_data))
print("Duplicate records:", len(duplicate_data))
print("Unique records:", len(unique_data))
```

```python
connection = sqlite3.connect("data_redundancy_database.db")

print("Database connected successfully.")
```

```python
existing_data.to_sql(
    "users",
    connection,
    if_exists="replace",
    index=False
)

print("Existing data stored in database.")
```

```python
database_data = pd.read_sql_query(
    "SELECT * FROM users",
    connection
)

database_data
```

```python
unique_data.to_sql(
    "users",
    connection,
    if_exists="append",
    index=False
)

print("Only unique and verified records added successfully.")
```

```python
final_data = pd.read_sql_query(
    "SELECT * FROM users",
    connection
)

final_data
```

```python
duplicate_check = final_data[final_data.duplicated(
    subset=["email"],
    keep=False
)]

if duplicate_check.empty:
    print("SUCCESS: No duplicate email records found.")
else:
    print("Duplicate records found:")
    print(duplicate_check)
```

```python
initial_count = len(existing_data)
incoming_count = len(new_data)
duplicate_count = len(duplicate_data)
unique_count = len(unique_data)
final_count = len(final_data)

print("========== DATA REDUNDANCY REPORT ==========")
print("Initial database records :", initial_count)
print("Incoming records         :", incoming_count)
print("Duplicate records        :", duplicate_count)
print("Unique records added     :", unique_count)
print("Final database records   :", final_count)
print("============================================")
```

```python
connection.close()

print("Database connection closed.")
```

```python
existing_data.to_csv("existing_data.csv", index=False)
new_data.to_csv("new_data.csv", index=False)
unique_data.to_csv("unique_verified_data.csv", index=False)
duplicate_data.to_csv("duplicate_data.csv", index=False)

print("CSV files saved successfully.")
```
